# Voting Analysis - Riksdagen API

This notebook fetches voting data from the Swedish Parliament's API and analyzes how different parties voted on a specific issue.

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_theme(style="whitegrid")
```

```python
# Define API Endpoint
API_BASE = "https://data.riksdagen.se/voteringlista/"

# Parameters for a specific vote (Example: 2023/24, Betänkande MJU18)
# You can change these to explore other votes
params = {
    "rm": "2023/24",
    "bet": "MJU18",
    "punkt": "1",
    "utformat": "json"
}
```

```python
# Fetch data
response = requests.get(API_BASE, params=params)

if response.status_code == 200:
    data = response.json()
    print("Data fetched successfully!")
else:
    print(f"Failed to fetch data. Status code: {response.status_code}")
```

```python
# Extract the list of votes
votes_list = data['voteringlista']['votering']

# Create DataFrame
df = pd.DataFrame(votes_list)

# Display first few rows
df[['namn', 'parti', 'rost', 'valkrets']].head()
```

```python
# Aggregate votes by party and vote type
party_votes = df.groupby(['parti', 'rost']).size().unstack(fill_value=0)

# Reorder columns for better readability if they exist
desired_order = ['Ja', 'Nej', 'Avstår', 'Frånvarande']
existing_columns = [col for col in desired_order if col in party_votes.columns]
party_votes = party_votes[existing_columns]

party_votes
```

```python
# Plot
party_votes.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='viridis')

plt.title('Voting Distribution by Party')
plt.xlabel('Party')
plt.ylabel('Number of Votes')
plt.legend(title='Vote')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
